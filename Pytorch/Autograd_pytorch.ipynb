```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)
```

```python
w=torch.tensor(1.0,requires_grad=True)
```

```python
b=torch.tensor(0.0,requires_grad=True)
```

```python
z= w*x+b
```

```python
y_pred = torch.sigmoid(z)
```

```python
loss = torch.binary_cross_entropy_with_logits(y_pred,y)
```

```python
loss.backward()
```

```python
print(w.grad,b.grad)
```
